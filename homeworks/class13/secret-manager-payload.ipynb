```python
import json
import boto3
client = boto3.client("secretsmanager")
```

```python
secret_id = "random-passwd"
```

```python
def get_password():
    response = client.get_secret_value(SecretId=secret_id)
    secret = json.loads(response["SecretString"])
    password = secret["password"]
    return password
    
```
